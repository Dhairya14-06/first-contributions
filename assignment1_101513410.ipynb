```python
"""
Author: Dhairya Gohel
Assignment: #1
"""

# String: Stores the name of the gym member
gym_member = "Alex Alliton"

# Float: Represents the preferred weight in kilograms
preferred_weight_kg = 20.5

# Integer: Represents the highest number of repetitions
highest_reps = 25

# Boolean: Indicates whether the membership is active or not
membership_active = True

# Dictionary: Stores friends' workout times in tuples (yoga, running, weightlifting)
workout_stats = {
    "Alex": (30, 45, 20),
    "Jamie": (40, 50, 30),
    "Taylor": (25, 35, 40)
}

# Temporary dictionary to store the new key-value pairs
updated_stats = {}

# Loop through workout_stats to calculate total minutes
for friend, times in workout_stats.items():
    total_minutes = sum(times)
    updated_stats[f"{friend}_Total"] = total_minutes

# Update the main dictionary after iteration
workout_stats.update(updated_stats)

# List of lists: Extracts workout times for each friend into a structured format
workout_list = [list(times) for friend, times in workout_stats.items() if "_Total" not in friend]

# Extract and print yoga and running minutes for all friends
yoga_running = [row[:2] for row in workout_list]
print("Yoga & Running Minutes:", yoga_running)

# Extract and print weightlifting minutes for the last two friends
weightlifting = [row[2] for row in workout_list[-2:]]
print("Weightlifting Minutes for last two friends:", weightlifting)

# Check if any friend has total workout time >= 120 minutes
for friend, total in updated_stats.items():
    if total >= 120:
        print(f"Great job staying active, {friend.replace('_Total', '')}!")

# Allow the user to enter a friend's name and display workout stats
friend_name = input("Enter a friend's name to check workout stats: ")

if friend_name in workout_stats:
    activities = ["Yoga", "Running", "Weightlifting"]
    stats = workout_stats[friend_name]
    print(f"{friend_name}'s workout details:")
    for activity, time in zip(activities, stats):
        print(f"{activity}: {time} minutes")
    print(f"Total Workout Minutes: {workout_stats[f'{friend_name}_Total']}")
else:
    print(f"Friend {friend_name} not found in the records.")

# Find highest and lowest workout totals
friends_total = {friend: total for friend, total in updated_stats.items()}

max_friend = max(friends_total, key=friends_total.get).replace("_Total", "")
min_friend = min(friends_total, key=friends_total.get).replace("_Total", "")

print(f"The friend with the highest total workout minutes: {max_friend}")
print(f"The friend with the lowest total workout minutes: {min_friend}")
```
